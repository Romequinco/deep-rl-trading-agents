## 1. Tabla de rendimiento

Mediana del retorno en test. Las filas de los tabulares vienen del Taller 1, con las mismas series de test.

```python
tabular = pd.read_csv("../results/taller1_tabular.csv" if Path.cwd().name == "notebooks" else "results/taller1_tabular.csv")
tabular = tabular[tabular.agente.isin(["qlearning", "sarsa"])]
resultados = pd.concat([load_results(), tabular], ignore_index=True)

def formato(r):
    return f"x{1 + r:.1f}" if r >= 5 else f"{r:+.1%}"

orden = [a for a in AGENT_COLORS if a in set(resultados.agente)]
tabla = resultados.pivot(index="agente", columns="mercado", values="retorno_mediana").loc[orden, list(MARKETS)]
tabla.map(formato).rename(columns=MARKET_LABELS)
```
